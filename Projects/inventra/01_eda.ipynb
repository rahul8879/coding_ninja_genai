```python
# clean the sales data
sales_data['date'] = pd.to_datetime(sales_data['date'])
sales_data['temperature'] = sales_data['temperature'].str.replace(' ', '').astype(float)
```

```python
sales_eda = sales_data.merge(inventory[['sku', 'name','category']], on='sku',
                              how='left')
```

```python
sales_eda.head()
```

```python
#q) does demand behave differently based on weather conditions?

weather_demand = sales_eda.groupby('weather_condition')['qty'].agg(["mean","median"]).sort_values(by="mean",
                                                                                  ascending=False)
```

```python
import matplotlib.pyplot as plt

```

```python
plt.figure(figsize=(10,5))
plt.bar(weather_demand.index, weather_demand['mean'], alpha=0.6, label='Mean Demand')
plt.bar(weather_demand.index, weather_demand['median'], alpha=0.6, label='Median Demand')
plt.title('Average Demand by Weather Condition')
plt.xlabel('Weather Condition')
plt.ylabel('Average Demand')
plt.xticks(rotation=45)
plt.legend()
plt.show()
```

```python
category_weather = pd.pivot_table(
    sales_eda,
    index = "category",
    columns = "weather_condition",
    values = "qty",
    aggfunc = "mean"
)

display(category_weather)
category_weather.plot(kind='bar', figsize=(30,15))
plt.title('Average Demand by Category and Weather Condition')
plt.xlabel('Category')
plt.ylabel('Average Demand')
plt.xticks(rotation=45)
plt.legend(title='Weather Condition')
plt.show()
```
