```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
fruits = ['apple', 'banana', 'cherry','lemon', 'orange','grape','mango']

# fruits ===> 2 d ===> properties ===? sweetness(0-10), size(0-10)

apple = [7, 5.0]
banana = [8, 9.0]
cherry = [6, 4.0]
lemon = [4, 3.0]
orange = [3, 7.0]
grape = [9, 2.0]
mango = [6, 8.0]


vs = [apple, banana, cherry, lemon, orange, grape, mango]

vs = {fruit: properties for fruit, properties in zip(fruits, vs)}
```

```python
def find_fruits(new_fruits_vector,vs,k):
    distances = {}
    for fruit, properties in vs.items():
        distance = np.sqrt((new_fruits_vector[0] - properties[0])**2 + (new_fruits_vector[1] - properties[1])**2)
        distances[fruit] = distance

    # sort it and return top k
    sorted_fruits = [x for _, x in sorted(zip(distances.values(), distances.keys()))]
    return sorted_fruits[:k]
```

```python
find_fruits([8,8], vs,k=2)
```

```python
# line 

not_spam = np.array([
    [1, 1],
    [1, 2],
    [2, 1],
    [2, 2],
    [3, 1],
])

spam = np.array([
    [6, 6],
    [7, 5],
    [6, 7],
    [8, 6],
    [7, 7],
])

a,b,c = 1,1,-6

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(not_spam[:, 0], not_spam[:, 1], color='blue')
ax.scatter(spam[:, 0], spam[:, 1], color='red')
x_values = np.linspace(0, 8, 100)
y_vals = (-a * x_values - c) / b
ax.plot(x_values, y_vals, color='green', label='Decision Boundary')
```

```python

a,b,c = 1,1,-6

def predict(vector):
    x = vector[0]
    y = vector[1]
    # Decision boundary: ax + by + c = 0
    if a * x + b * y + c > 0:
        return "not spam"
    else:
        return "spam"
    

query = np.array([2, 1])
predict(query)
```
